```python
class Student:
    def __init__(self,name,place):
        self.name=name
        self.place=place
    def show(self):
        return f'The name is {self.name} and you are from {self.place}'
```

```python
p1=Student('ABC' , 'Mysuru')
```

```python
p1.show()
```

```python
class student:
    def __init__(std,usn,name,place):
        std.usn=usn
        std.name=name
        std.place=place
    def show(std):
        print(f'USN of {std.usn} having the name {std.name} is from {std.place}')
    def update(std,usn,place):
        std.usn=usn
        std.place=place
    
```

```python
stud=student(130,'RMS','Mys')
stud.name
```

```python
stud.show()
```

```python
stud.update(19,'blr')
```

```python
stud.show()
```
